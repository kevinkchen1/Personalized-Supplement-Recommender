```python
import pandas as pd

# Read the CSV file
df = pd.read_csv('drugbank_data/drug_categories.csv')

# Get unique categories
unique_categories = df['category'].unique()

# Create category IDs in the format C00001, C00002, etc.
category_ids = [f'C{str(i+1).zfill(5)}' for i in range(len(unique_categories))]

# Create the mapping DataFrame
category_mapping = pd.DataFrame({
    'category_id': category_ids,
    'category': unique_categories
})

# Save to CSV
category_mapping.to_csv('categories.csv', index=False)

print(category_mapping)
```

```python
import pandas as pd

# Read both CSV files
drug_categories = pd.read_csv('drugbank_data/drug_categories.csv')
categories = pd.read_csv('drugbank_data/categories.csv')

# Merge the dataframes on the 'category' column
drug_category_belongs_to = drug_categories.merge(
    categories, 
    on='category', 
    how='left'
)

# Keep only the columns we need
drug_category_belongs_to = drug_category_belongs_to[['drug_id', 'category_id']]

# Save to CSV
drug_category_belongs_to.to_csv('drug_category_belongs_to.csv', index=False)

print(drug_category_belongs_to)
```

```python
import pandas as pd

# Read the CSV file
df = pd.read_csv('drugbank_data/drug_products.csv')  # Replace 'your_file.csv' with your actual filename

# Get unique combinations of product info (excluding drug_id)
unique_brands = df[['product_name', 'labeller', 'dosage_form', 'strength', 'route']].drop_duplicates()

# Create brand_name_id in the format B00001, B00002, etc.
unique_brands.insert(0, 'brand_name_id', [f'B{str(i+1).zfill(5)}' for i in range(len(unique_brands))])

# Rename product_name to brand_name
unique_brands = unique_brands.rename(columns={'product_name': 'brand_name'})

# Reset index
unique_brands = unique_brands.reset_index(drop=True)

# Save to CSV
unique_brands.to_csv('brand_names.csv', index=False)

print(unique_brands)
```

```python
import pandas as pd

# Read the CSV files
drug_products = pd.read_csv('drugbank_data/drug_products.csv')
brand_names = pd.read_csv('drugbank_data/brand_names.csv')

# Merge drug_products with brand_names to get brand_name_id
# Match on product_name = brand_name and all other attributes
merged = drug_products.merge(
    brand_names,
    left_on=['product_name', 'labeller', 'dosage_form', 'strength', 'route'],
    right_on=['brand_name', 'labeller', 'dosage_form', 'strength', 'route'],
    how='left'
)

# Keep only the columns we need
brand_name_drug_contains = merged[['brand_name_id', 'drug_id']]

# Remove duplicates (in case there are any)
brand_name_drug_contains = brand_name_drug_contains.drop_duplicates()

# Save to CSV
brand_name_drug_contains.to_csv('brand_name_drug_contains.csv', index=False)

print(brand_name_drug_contains)
```

```python
import pandas as pd

# Read the CSV file
df = pd.read_csv('drugbank_data/drug_synonyms.csv')

# Get unique synonyms
unique_synonyms = df['synonym'].unique()

# Create synonym IDs in the format SYN00001, SYN00002, etc.
synonym_ids = [f'SYN{str(i+1).zfill(5)}' for i in range(len(unique_synonyms))]

# Create the mapping DataFrame
synonym_mapping = pd.DataFrame({
    'synonym_id': synonym_ids,
    'synonym': unique_synonyms
})

# Save to CSV
synonym_mapping.to_csv('synonyms.csv', index=False)

print(synonym_mapping)
```

```python
import pandas as pd

# Read both CSV files
drug_synonyms = pd.read_csv('drugbank_data/drug_synonyms.csv')
synonyms = pd.read_csv('drugbank_data/synonyms.csv')

# Merge the dataframes on the 'synonym' column
drug_synonym_known_as = drug_synonyms.merge(
    synonyms, 
    on='synonym', 
    how='left'
)

# Rename drugbank_id to drug_id and keep only the columns we need
drug_synonym_known_as = drug_synonym_known_as.rename(columns={'drugbank_id': 'drug_id'})
drug_synonym_known_as = drug_synonym_known_as[['drug_id', 'synonym_id']]

# Save to CSV
drug_synonym_known_as.to_csv('drug_synonym_known_as.csv', index=False)

print(drug_synonym_known_as)
```

```python
import pandas as pd

# Read both CSV files
drugs = pd.read_csv('drugbank_data/drugs.csv')
toxicity_contraindications = pd.read_csv('drugbank_data/toxicity_contraindications.csv')

# Merge the dataframes on 'drug_id'
# Use left merge to keep all drugs, and fill in toxicity/indication_text where available
drugs_updated = drugs.merge(
    toxicity_contraindications[['drug_id', 'toxicity', 'indication_text']], 
    on='drug_id', 
    how='left'
)

# Save to CSV
drugs_updated.to_csv('drugs.csv', index=False)

print(drugs_updated)
```

```python
import pandas as pd

# Read both CSV files
drugs = pd.read_csv('drugbank_data/drugs.csv')
toxicity_contraindications = pd.read_csv('drugbank_data/toxicity_contraindications.csv')

# Remove duplicate drug_ids, keeping the last occurrence
toxicity_contraindications = toxicity_contraindications.drop_duplicates(subset='drug_id', keep='last')

# Merge the dataframes on 'drug_id'
drugs_updated = drugs.merge(
    toxicity_contraindications[['drug_id', 'toxicity', 'indication_text']], 
    on='drug_id', 
    how='left'
)

# Save to CSV
drugs_updated.to_csv('drugs.csv', index=False)

print(drugs_updated)
```
